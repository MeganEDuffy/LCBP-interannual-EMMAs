```r
library(fs)

# Set directories
home_dir <- Sys.getenv("HOME")
repo_dir <- file.path(home_dir, "OneDrive/git-repos/LCBP-interannual-EMMAs")

source(file.path(repo_dir, "Functions/calculate_tracer_rrmse_table.R"))

data_dir  <- file.path(repo_dir, "Data/")
chem_file <- file.path(data_dir, "GrabSample_data", "RI23-IC-ICP-isotope-toc-joined.csv")

# 1. Generate Table 3 for Hungerford Brook (1-6 dimensions)

hungerford_tracers <- c("Mn_mg_L", "Ca_mg_L", "Mg_mg_L", "Na_mg_L", "Cl_mg_L", "Si_mg_L", "dD", "d18O", "SO4_mg_L", "K_mg_L", "Cu_mg_L", "P_mg_L")

hungerford_table3 <- calculate_tracer_rrmse_table(
  chem_data_file   = chem_file,
  site_name        = "Hungerford",
  max_dims         = 6,
  selected_tracers = hungerford_tracers
)
print("=== Table S2: Hungerford Brook RRMSE Projection of streamwater observations in a mixing space created by its own eigenvectors (1-6) ===")
print(hungerford_table3)

# 2. Generate Table 3 for Wade Brook (1-6 dimensions)

wade_tracers <- c("Mn_mg_L","Cu_mg_L", "P_mg_L", "Cl_mg_L", "SO4_mg_L", "Ca_mg_L", "Mg_mg_L", "Na_mg_L", "dD", "d18O", "DOC_mg_L", "K_mg_L")

wade_table3 <- calculate_tracer_rrmse_table(
  chem_data_file   = chem_file,
  site_name        = "Wade",
  max_dims         = 6,
  selected_tracers = wade_tracers
)
print("=== Table S3: Wade Brook RRMSE Projection of streamwater observations in a mixing space created by its own eigenvectors (1-6) ===")
print(wade_table3)
```
