```r
library(fs)
library(ggplot2)
library(GGally)

# Set directories
home_dir <- Sys.getenv("HOME")
repo_dir <- file.path(home_dir, "OneDrive/git-repos/LCBP-interannual-EMMAs")
source(file.path(repo_dir, "Functions/plot_tracer_bivariate_grid.R"))

data_dir   <- file.path(repo_dir, "Data/")
output_dir <- file.path(repo_dir, "Output/Tracer-selection/")
chem_file  <- file.path(data_dir, "GrabSample_data", "RI23-IC-ICP-isotope-toc-joined.csv")

# 1. Run for Hungerford Brook
hungerford_results <- plot_tracer_bivariate_grid(
  chem_data_file = chem_file,
  site_name      = "Hungerford",
  output_dir     = output_dir,
  base_font_size = 10
)

print(hungerford_results$plot)
cat("Hungerford Streamwater Sample Count (n):", hungerford_results$n_samples, "\n")

# 2. Run for Wade Brook
wade_results <- plot_tracer_bivariate_grid(
  chem_data_file = chem_file,
  site_name      = "Wade",
  output_dir     = output_dir,
  base_font_size = 10
)

print(wade_results$plot)
cat("Wade Streamwater Sample Count (n):", wade_results$n_samples, "\n")
```
