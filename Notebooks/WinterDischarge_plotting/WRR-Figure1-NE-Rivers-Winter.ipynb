```r
library(fs)
library(ggplot2)
library(cowplot)
library(dplyr)

# Set directories
home_dir <- Sys.getenv("HOME")
repo_dir <- file.path(home_dir, "OneDrive/git-repos/LCBP-interannual-EMMAs")
source(file.path(repo_dir, "Functions/plot_cumulative_winter_discharge.R"))

output_dir <- file.path(repo_dir, "Output/Event-yields/")

# 1. Generate individual river results (each returns a list)
res1 <- plot_cumulative_winter_discharge(
  site_number  = "USGS-01184000",
  site_name    = "Connecticut River", # at Thompsonville
  panel_label  = "a)", 
  start_date   = "1929-10-01",
  end_date     = "2025-01-03"
)

res2 <- plot_cumulative_winter_discharge(
  site_number  = "USGS-01100000",
  site_name    = "Merrimack River", # at Lowell, MA
  panel_label  = "b)", 
  start_date   = "1929-10-01",
  end_date     = "2025-01-03"
)

res3 <- plot_cumulative_winter_discharge(
  site_number  = "USGS-04290500",
  site_name    = "Winooski River",
  panel_label  = "c)", 
  start_date   = "1929-10-01",
  end_date     = "2025-01-03"
)

res4 <- plot_cumulative_winter_discharge(
  site_number  = "USGS-04292500",
  site_name    = "Lamoille River",
  panel_label  = "d)", 
  start_date   = "1929-10-01",
  end_date     = "2025-01-03"
)

# 2. Extract ggplot objects from the lists
p1 <- res1$plot
p2 <- res2$plot
p3 <- res3$plot
p4 <- res4$plot

# Combine stats tables into a single summary data frame
all_stats <- bind_rows(res1$stats, res2$stats, res3$stats, res4$stats)

# 3. Adjust themes and combine plots
p1 <- p1 + theme(legend.position = "none")
p2 <- p2 + theme(legend.position = "none")
p3 <- p3 + theme(legend.position = "none")

grid_composite <- plot_grid(
  p1, p2, p3, p4,
  ncol = 2,
  nrow = 2,
  align = "hv"
)

# 4. Save composite figure & summary table
dir.create(output_dir, showWarnings = FALSE, recursive = TRUE)

file_path <- file.path(output_dir, "Duffy_Fig1_Four_Rivers_Cumulative_Winter_Discharge_2x2.jpg")
ggsave(plot = grid_composite, filename = file_path, width = 12, height = 11, dpi = 300)

stats_path <- file.path(output_dir, "Four_Rivers_Winter_Discharge_Stats.csv")
write.csv(all_stats, stats_path, row.names = FALSE)

print(grid_composite)
```
