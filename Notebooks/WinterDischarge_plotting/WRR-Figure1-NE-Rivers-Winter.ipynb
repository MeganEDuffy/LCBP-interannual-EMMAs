```r
library(fs)
library(ggplot2)
library(cowplot)

# Set directories
home_dir <- Sys.getenv("HOME")
repo_dir <- file.path(home_dir, "OneDrive/git-repos/LCBP-interannual-EMMAs")
source(file.path(repo_dir, "Functions/plot_cumulative_winter_discharge.R"))

output_dir <- file.path(repo_dir, "Output/Event-yields/")

# 1. Generate individual river plots using the updated function
p1 <- plot_cumulative_winter_discharge(
  site_number  = "USGS-01184000",
  site_name    = "Connecticut River", #at Thompsonville
  panel_label  = "a)", 
  start_date   = "1929-10-01",
  end_date     = "2025-01-03"
)

p2 <- plot_cumulative_winter_discharge(
  site_number  = "USGS-01100000",
  site_name    = "Merrimack River", # at Lowell, MA
  panel_label  = "b)", 
  start_date   = "1929-10-01",
  end_date     = "2025-01-03"
)

p3 <- plot_cumulative_winter_discharge(
  site_number  = "USGS-04290500",
  site_name    = "Winooski River",
  panel_label  = "c)", 
  start_date   = "1929-10-01",
  end_date     = "2025-01-03"
)

p4 <- plot_cumulative_winter_discharge(
  site_number  = "USGS-04292500",
  site_name    = "Lamoille River",
  panel_label  = "d)", 
  start_date   = "1929-10-01",
  end_date     = "2025-01-03"
)

# 2. Combine into a 2x2 Grid Layout using cowplot

# Only keep the legend for one plot (e.g., p4)
p1 <- p1 + theme(legend.position = "none")
p2 <- p2 + theme(legend.position = "none")
p3 <- p3 + theme(legend.position = "none")

# Use cowplot to stitch them
grid_composite <- plot_grid(p1, p2, p3, p4, ncol = 2)
grid_composite <- plot_grid(
  p1, p2, p3, p4,
  ncol = 2,
  nrow = 2,
  align = "hv"
)

# 3. Save the 2x2 composite figure
dir.create(output_dir, showWarnings = FALSE, recursive = TRUE)
file_path <- file.path(output_dir, "Four_Rivers_Cumulative_Winter_Discharge_2x2.jpg")

ggsave(plot = grid_composite, width = 11, height = 10, dpi = 300, file = file_path)

print(grid_composite)
```
