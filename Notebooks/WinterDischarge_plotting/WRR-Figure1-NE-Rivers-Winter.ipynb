```r
library(fs)
library(ggplot2)

# Set directories
home_dir <- Sys.getenv("HOME")
repo_dir <- file.path(home_dir, "OneDrive/git-repos/LCBP-interannual-EMMAs")
source(file.path(repo_dir, "Functions/plot_cumulative_winter_discharge.R"))

output_dir <- file.path(repo_dir, "Output/Event-yields/")

# Run the function for the Connecticut River at Thompsonville
connecticut_cumulative <- plot_cumulative_winter_discharge(
  site_number  = "USGS-01184000",
  site_name    = "Connecticut River at Thompsonville",
  panel_label = "a)", 
  output_dir   = output_dir,
  start_date   = "1929-10-01",
  end_date     = "2025-01-03"
)

print(connecticut_cumulative)
```
